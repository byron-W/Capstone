```python
import pandas as pd
```

```python
weather_data = pd.read_csv('./data/real/weather_data.csv')
last_temp = pd.read_csv('./data/real/3rdyr.csv')
last_hi = pd.read_csv('./data/real/3yrHI.csv')
last_p = pd.read_csv('./data/real/3yrP.csv')
last_V = pd.read_csv('./data/real/3yrV.csv')
last_W = pd.read_csv('./data/real/3yrW.csv')
last_WC = pd.read_csv('./data/real/3yrWC.csv')
pd.set_option('display.max_columns', None)
weather_data.tail()
```

```python
last_temp.head()
```

```python
last_temp['One Hour Precipitation'] = last_p['One Hour Precipitation']
last_temp['Heat Index'] = last_hi['Heat Index']
last_temp['Wind Chill'] = last_WC['Wind Chill']
last_temp['Wind Speed'] = last_W['Wind Speed']
last_temp['Gust'] = last_W['Gust']
last_temp['Direction'] = last_W['Direction']
last_temp['Visibility'] = last_V['Visibility']

last_temp.head()
```

```python
weather = pd.concat((weather_data, last_temp))
print(len(weather_data), len(last_temp), len(weather), (len(weather_data) + len(last_temp)))
weather.head()
```

```python
weather.tail()
```

```python
weather['DateTime'] = pd.to_datetime(weather['DateTime'], format='mixed')
weather['Hour'] = weather['DateTime'].dt.hour
weather['Date'] = weather['DateTime'].dt.date
weather = weather[['DateTime', 'Date', 'Hour', 'Temperature', 'Dew Point', 'Relative Humidity', 'One Hour Precipitation', 'Heat Index', 'Wind Chill', 'Wind Speed', 'Gust', 'Direction', 'Visibility']]
weather = weather.reset_index().drop('index', axis=1)
weather = weather.sort_values(['Date', 'Hour'])
weather.head()
```

```python
hour = weather['Hour'] == 19
date = weather['Date'] == pd.Timestamp('2025-07-31').date()
display(weather.loc[hour& date])
print(len(weather))
```

```python
# weather['DateTime'] = pd.to_datetime(weather['Date'].astype(str) + ' ' + weather['Hour'].astype(int).astype(str) + ':00')

# Define start and end times
start_time = pd.Timestamp('2022-07-31 20:00:00')
end_time = pd.Timestamp('2025-07-31 19:00:00')

start_time = weather.iloc[0]['DateTime']
end_time = weather.iloc[-1]['DateTime']
#end_time = pd.Timestamp('2025-07-31 19:00:00')
# Generate full expected hourly range
expected_times = pd.date_range(start=start_time, end=end_time, freq='h')

# Identify which timestamps are missing
actual_times = set(weather['DateTime'])
missing_times = [t for t in expected_times if t not in actual_times]

# Display number of missing timestamps and a sample (if any)
len(missing_times), missing_times[:10]
```

```python
weather = weather.drop_duplicates(subset='DateTime', keep='first')

weather = weather.set_index('DateTime').reindex(expected_times)

# Restore Date and Hour columns
weather['Date'] = weather.index.date
weather['Hour'] = weather.index.hour
weather = weather.reset_index(drop=True)
```

```python
internet = pd.read_csv('./data/real/internet.csv')
internet.head()
```

```python
internet['date'] = pd.to_datetime(internet['date'], format='mixed')
internet = internet.rename({
    'date' : 'Date',
    'hour' : 'Hour',
    'avg_download_speed_mbps' : 'Download Speed',
    'num_tests' : 'Test Count'
    }, axis=1)
internet = internet.dropna(subset=['Hour'])
internet['Hour'] = internet['Hour'].astype(int)
print(len(internet), len(weather))
internet.head()
```

```python
for df in [weather, internet]:
    df['Date'] = pd.to_datetime(df['Date']).dt.date
    df['Hour'] = pd.to_numeric(df['Hour']).astype(int)
```

```python
weather.to_csv('./data/real/all_weather_data.csv')
internet.to_csv('./data/real/all_internet_data.csv')
```
